# FASE 01: ENTENDIMENTO DO NEGÓCIO

📖 **O que é:** antes de qualquer dado, entender **qual problema resolver, para quem e como saber se deu certo**. Um modelo excelente para a pergunta errada não tem valor.

## 🏢 Contexto

📖 Descreva o problema: o catálogo é enorme, mas cada cliente vê poucos itens na tela. Mostrar a todos os mesmos best-sellers é fácil, mas deixa de fora o que **cada pessoa** gostaria.

✅ Inclua: a fonte dos dados (quem consumiu o quê, com ou sem nota), o **stakeholder** (ex.: o time da vitrine do site) e onde a recomendação aparece.

## 🎯 Objetivo

📖 "Recomendar a cada leitor **N itens** que ele ainda não consumiu e de que provavelmente vai gostar, e **medir o acerto** contra a vitrine de mais populares."

## 🧩 Formato da solução

| Item | Exemplo |
|---|---|
| **Granularidade** | Um usuário (ou um leitor novo que informa alguns itens de que gostou) |
| **Tipo de problema** | Recomendação top-N por **filtragem colaborativa** (aprende "quem consumiu o quê") |
| **O que conta como acerto** | Um item que o usuário avaliou bem (ex.: nota ≥ 7), **escondido do modelo**, aparecer entre os N recomendados |
| **Métrica principal** | Acerto@N, comparado com a popularidade e o sorteio |
| **Entrega** | Modelo salvo, ficha e app com as recomendações, as capas e o motivo de cada uma |

## 🗺️ Planejamento da solução (CRISP-DM)

**CRISP-DM** (*Cross Industry Standard Process for Data Mining*) é um roteiro de 6 fases para projetos de dados. Ele é **cíclico**: se a avaliação mostrar que o modelo não responde à pergunta de negócio, voltamos ao entendimento do negócio ou aos dados, com o que aprendemos. A vantagem é entregar valor rápido e ir melhorando a cada volta.

![Ciclo CRISP-DM](imagens/crisp_dm.svg)

| Fase | Pergunta que responde |
|---|---|
| **01. Entendimento do negócio** | Qual problema queremos resolver e como saberemos que deu certo? |
| **02. Preparação dos dados** | Os dados são confiáveis? O que eles já contam (EDA)? |
| **03. Modelagem** | Qual método resolve melhor, medido de forma honesta? |
| **04. Avaliação** | O resultado vale para o negócio? Onde o modelo erra? |
| **05. Implantação** | Como as pessoas vão usar (modelo salvo, app, API)? |
| **Insights e próximos passos** | O que aprendemos e o que fazer na próxima volta? |

### 🧠 Os tipos de aprendizado de máquina

| Tipo | O que o modelo recebe | Exemplos de problema |
|---|---|---|
| **Supervisionado** | Exemplos **com a resposta certa** (o alvo) | Classificação (vai sair? é fraude?) e regressão (quanto vai vender?) |
| **Não supervisionado** | Dados **sem resposta**; o modelo procura padrões sozinho | Clusterização (agrupamento, que é o mesmo nome) e redução de dimensionalidade (PCA) |
| **Semi-supervisionado** | Poucos exemplos com resposta e muitos sem | Rotular milhões de fotos tendo só algumas rotuladas à mão |
| **Por reforço** | Uma recompensa ou punição a cada ação | Robôs, jogos, precificação dinâmica |

> Recomendação (sugerir livros, filmes) usa muitas vezes aprendizado sem alvo explícito: o modelo aprende "quem consumiu o quê" e é avaliado escondendo parte do histórico.

✅ **O que fazer:** ler os arquivos, mostrar as primeiras linhas (`.head()`) e conferir se a leitura está certa (separador, acentos, tipos).

### 2.2. Colunas

📖 **O que é:** entender o **significado real** de cada coluna e dar nomes claros, em português e sem espaços (`snake_case`).

✅ **O que fazer**
- Montar um **dicionário de dados**: coluna original → nome novo → significado → unidade.

### 2.3. Dimensões dos dados

✅ **O que fazer**
- Número de linhas e colunas (`.shape`) e de valores distintos por coluna (`.nunique()`).
- **Duplicatas:** linhas repetidas inflam as métricas quando uma cópia cai no treino e outra no teste (no projeto de Sentimento, 23% da base era duplicada).
- Chaves que deveriam bater entre tabelas (ex.: 11,5% das notas dos Livros eram de livros fora do cadastro).

### 2.4. Valores nulos

📖 **O que é:** valor nulo (`NaN`, *missing value*) é um dado que **não foi registrado**. Ele pode vir de dado corrompido, truncado (só parte chegou), formulário incompleto ou de algo que **não se aplica** (ex.: "data da demissão" de quem ainda trabalha).

🧠 **Por que o motivo importa**

| Tipo de ausência | Exemplo | Consequência |
|---|---|---|
| **Ao acaso** | Falha de sistema apagou algumas linhas | Preencher com a mediana é seguro |
| **Ligada a outra coluna** | Clientes novos não têm "tempo de casa" | A outra coluna explica; preencha por grupo |
| **Ligada ao próprio valor** | Quem ganha muito não informa a renda | O nulo **é informação**: crie uma coluna "renda_informada" |

✅ **Formas de tratar**
1. **Remover as linhas:** só quando são poucas e a ausência é ao acaso.
2. **Preencher (imputar)** com a mediana (mais segura que a média quando há valores extremos), a moda (categóricas) ou um valor que faça sentido no negócio (0 compras = não comprou).
3. **Entender o porquê** e, se o nulo carrega informação, criar um **indicador de ausência**.

⚠️ **Erros comuns:** calcular a mediana com a base inteira vaza informação do teste para o treino. Faça a imputação **dentro do Pipeline** (`SimpleImputer`), que aprende a mediana só no treino.

### 2.5. Estatística descritiva

📖 **O que é:** resumir cada coluna com poucos números e gráficos. Serve para duas coisas: **ganhar conhecimento de negócio** e **detectar erros** (idade de 244 anos, ano 2050, preço negativo). Separe as colunas em **numéricas** e **categóricas**, porque cada tipo tem suas ferramentas.

🧠 **Tendência central (onde fica o "meio" dos dados)**

| Medida | O que é | Exemplo: salários 2, 2, 3, 4 e 50 mil |
|---|---|---|
| **Média** | Soma ÷ quantidade | 12,2 mil (puxada pelo salário de 50) |
| **Mediana** | O valor do meio, com os dados em ordem | 3 mil (não se abala com o 50) |
| **Moda** | O valor mais frequente | 2 mil |

🧠 **Dispersão (o quanto os dados se espalham)**
- **Mínimo, máximo e amplitude** (máximo − mínimo).
- **Desvio padrão:** distância típica dos valores até a média (é a raiz quadrada da variância). Desvio pequeno = valores parecidos.
- **Quartis:** Q1 (25% dos dados abaixo), Q2 = mediana (50%), Q3 (75%). O **IQR** (Q3 − Q1) é o tamanho da metade central.

🧠 **Forma da distribuição**
- **Assimetria (*skew*):** para que lado fica a **cauda longa**.
  - **Positiva** (skew > 0): cauda à direita, poucos valores muito altos puxam a **média para cima da mediana** (renda, preço de imóveis, compras).
  - **Zero:** simétrica, média ≈ mediana.
  - **Negativa** (skew < 0): cauda à esquerda, média abaixo da mediana (nota de uma prova fácil).
- **Curtose:** o "peso" das caudas. Curtose alta = valores extremos aparecem mais do que numa curva normal.

![Assimetria](imagens/assimetria.svg)

⚠️ **Correção de uma ideia comum:** "o modelo precisa que os dados sigam uma distribuição normal". Não é bem assim. Modelos de árvore (Random Forest, XGBoost, LightGBM) não exigem normalidade. Modelos lineares funcionam melhor quando os **erros** são simétricos, e por isso o **log** ajuda em alvos com cauda longa (aluguel, vendas).

🧠 **Como ler um histograma**

O histograma divide os valores em **faixas** (as barras) e mostra **quantos dados caem em cada faixa** (a altura). Olhe quatro coisas: onde fica o **pico** (o valor mais comum), se há **mais de um pico** (grupos diferentes misturados), para que lado vai a **cauda** e se há **barras isoladas** longe do resto (outliers ou erros).

![Como ler um histograma](imagens/histograma_como_ler.svg)

🧠 **Outros gráficos**
- **Boxplot:** mostra mediana, quartis, IQR e outliers numa figura só (veja o item 2.6).
- **Gráfico de dispersão (*scatter*):** um ponto por linha, com uma variável em cada eixo; mostra se duas variáveis andam juntas (item 5).
- **Barras:** contagem de cada categoria.

✅ **O que fazer:** `describe()` das numéricas, contagem das categóricas e os **valores mais frequentes** de cada coluna, não só média e máximo.

⚠️ **Erro comum (Aluguel):** R$ 15.000 era o 3º aluguel mais comum, com só 12 anúncios acima: era o **teto da coleta**, não um preço real. Só a contagem de valores mostrou isso.

### 2.6. Outliers

📖 **O que é:** outlier (valor extremo, atípico) é uma observação muito afastada das demais. Ele pode ser:
- **erro de digitação ou de coleta** (idade 244): corrigir ou remover;
- **erro de medição** (sensor com defeito): corrigir ou remover;
- **natural** (um cliente que realmente compra 50 vezes mais): **manter**, porque é real, e tratar com escala robusta, log ou modelos de árvore.

🧠 **Regra do IQR (padrão do portfólio, função `regra_iqr`)**

São outliers os valores abaixo de **Q1 − 1,5 × IQR** ou acima de **Q3 + 1,5 × IQR**. Exemplo: Q1 = 40, Q3 = 60 → IQR = 20 → limites 10 e 90; um valor de 95 é outlier.

![Boxplot e regra do IQR](imagens/boxplot_iqr.svg)

🧠 **Outras técnicas**
- **Biblioteca PyOD:** algoritmos que olham várias colunas ao mesmo tempo e dizem **qual linha** é atípica (útil quando nenhuma coluna isolada parece estranha).
- **Teste de Grubbs:** testa se o valor mais extremo é outlier, mas **supõe dados quase normais**; em dados assimétricos, acusa outliers demais. Prefira o IQR.

✅ **O que fazer:** aplicar a regra do IQR em cada numérica, mostrar a % de outliers e escrever a **decisão** (manter, corrigir ou tratar) com o motivo.

### 2.7. Balanceamento do alvo

📖 Não há classes a equilibrar, mas olhe a **distribuição das interações**: quantas têm nota, quantas são "gostou" e quão concentrada é a atividade (no projeto de Livros, os 10% de usuários mais ativos faziam 81,6% das interações).

⚠️ **Erro comum (Livros):** a nota 0 era **interação sem nota**, não nota ruim. Descartá-la jogou fora 62% da informação, e usá-la foi o maior ganho do modelo.

### 🛡️ Remoção de variáveis com vazamento (*data leakage*)

📖 Num recomendador, vazamento é o modelo **ver os itens que deveria adivinhar**. A proteção está na separação (Fase 03, item 4.1): os itens escondidos de cada usuário não entram no treino, nem como interação nem pela nota. A nota só define **o que conta como acerto**. Atributos como a nota média de um item (para um filtro de qualidade) devem ser calculados **só com o treino**.

## 3 - ANÁLISE EXPLORATÓRIA DOS DADOS

📖 **O que é:** a EDA (*Exploratory Data Analysis*) mede o **impacto das variáveis** e começa a gerar valor antes de qualquer modelo. Ela tem três objetivos:
1. **entender como o negócio funciona** através dos dados;
2. **validar hipóteses**, que é onde surgem os *insights*. Um bom insight **surpreende** (informação que as pessoas não sabiam) ou **contraria uma crença** (o que todos achavam se mostra falso);
3. **ganhar sensibilidade** sobre quais variáveis influenciam o fenômeno, o que guia a modelagem.

Cada hipótese é uma afirmação testável ("clientes que sacam no cartão pagam menos a fatura inteira") e recebe um **teste estatístico com tamanho do efeito**.

### 🧪 Testes estatísticos: guia rápido

#### 1. A ideia por trás de qualquer teste

Imagine que você joga uma moeda 100 vezes e sai cara 60 vezes. A moeda é viciada ou foi sorte?

- **Hipótese nula (H0):** a hipótese "sem graça", de que **não há efeito** ("a moeda é honesta", "os grupos são iguais", "não há relação").
- **Hipótese alternativa (H1):** o que queremos mostrar ("a moeda é viciada").
- **p-valor:** a chance de ver um resultado **pelo menos tão extremo quanto o observado, SE H0 fosse verdade**. Com a moeda honesta, 60 caras ou mais (ou 40 ou menos) acontecem em cerca de 5,7% das vezes: p ≈ 0,057.
- **Nível de significância (α):** o corte combinado antes, em geral **0,05**. Se p < α, dizemos que o resultado é "estatisticamente significativo" e rejeitamos H0.
- **Erro tipo I (alarme falso):** rejeitar H0 quando ela era verdadeira (acusar a moeda honesta). Acontece em até α das vezes.
- **Erro tipo II (deixar passar):** não rejeitar H0 quando havia efeito (não perceber a moeda viciada). É mais comum com amostras pequenas.
- **Intervalo de confiança (IC 95%):** a faixa de valores plausíveis para o efeito. Se ele inclui o "sem efeito" (0, ou 1 no caso de odds ratio), o resultado é inconclusivo.

![p-valor](imagens/p_valor.svg)

⚠️ **O p-valor NÃO é** a chance de H0 ser verdadeira, nem o tamanho do efeito. Um p minúsculo diz que o efeito provavelmente existe, **não que ele é grande ou importante**.

#### 2. O que importa é o tamanho do efeito, não o p

Com muitos dados, **qualquer diferença minúscula vira "significativa"**. O tamanho do efeito responde à pergunta que importa para o negócio: **quanto?**

![Efeito × amostra](imagens/efeito_vs_amostra.svg)

Na figura, com 20.000 pessoas por grupo, uma diferença desprezível (d de Cohen = 0,10) dá p = 2e-23, enquanto, com 20 pessoas por grupo, um efeito grande (d = 1,00) dá p = 0,003. Na Eleição e nos Livros, várias hipóteses tiveram p quase zero e efeito desprezível. Por isso, toda conclusão traz a frase: **"o que importa é o tamanho do efeito, não o p"**.

#### 3. Qual teste usar

![Qual teste usar](imagens/qual_teste.svg)

| Situação | Teste | Tamanho do efeito (pequeno / médio / grande) | Exemplo do portfólio |
|---|---|---|---|
| Duas categóricas | **Qui-quadrado** | **V de Cramér** (0,1 / 0,3 / 0,5) | Rotatividade: departamento × saiu da empresa |
| Duas categóricas sim/não: "quantas vezes mais chance?" | **Odds ratio** | OR (1 = sem efeito; 2 = o dobro da chance) | Livros: quem leu Harry Potter 1 tem 268 vezes mais chance de ler o 2 |
| Uma numérica em 2 grupos, sem supor normalidade | **Mann-Whitney** | **Correlação de postos** (0,1 / 0,3 / 0,5) | Segmentação: quem saca no cartão × % da fatura paga |
| Uma numérica em 3 ou mais grupos | **Kruskal-Wallis** | **Épsilon²** (0,01 / 0,08 / 0,26) | Segmentação: diferença de cada variável entre os perfis |
| Duas numéricas que sobem ou descem juntas | **Spearman** | **rho** (0,1 / 0,3 / 0,5) | Livros: número de notas × nota média |
| Duas numéricas com relação em linha reta e sem outliers | **Pearson** | **r** (0,1 / 0,3 / 0,5) | (em dados assimétricos, prefira o Spearman) |
| A mesma unidade medida duas vezes | **Wilcoxon pareado** | Mediana da diferença ou correlação de postos pareada | Segmentação: compras à vista × parceladas do mesmo cliente |
| Uma proporção contra um valor de referência | **Binomial** | A própria proporção e o IC | Livros: 62,3% das interações sem nota (contra 50%) |
| Médias de 2 grupos com dados quase normais | **t de Welch** | **d de Cohen** (0,2 / 0,5 / 0,8) | Raramente: dados de negócio costumam ser assimétricos |

> 🧠 **Paramétrico × não paramétrico:** testes como t e Pearson supõem uma forma para os dados (normal, sem outliers). Os testes de **postos** (Mann-Whitney, Kruskal, Spearman, Wilcoxon) trocam os valores pela posição na fila (1º, 2º, 3º…) e funcionam com dados assimétricos e com outliers. Por isso são o padrão do portfólio.

#### 4. Como escrever a conclusão

```
✅ **H5 confirmada:** quem saca no cartão paga a fatura inteira com menos frequência
(mediana de 0% × 8%).

🧪 **Teste estatístico:** Mann-Whitney, correlação de postos de **−0,29** (efeito
moderado), p < 0,001.
```

❌ Ruim: "p < 0,05, então a variável é importante" (não diz o quanto, nem o sentido).

⚠️ **Cuidados**
- **Muitos testes ao mesmo tempo:** com 20 testes e α = 0,05, espera-se cerca de 1 "significativo" por puro acaso. Desconfie de um achado isolado.
- **Poucas observações:** o teste perde força (erro tipo II). Com 5 campanhas ou 6 ações, escreva que o resultado é frágil.
- **Correlação não é causalidade:** vendas de sorvete e afogamentos sobem juntos por causa do verão, não um por causa do outro.
- **Bootstrap:** quando não existe fórmula para a incerteza (ex.: índice de Sharpe), reamostre os dados milhares de vezes e veja a faixa do resultado (IC). Se o IC inclui o zero, não dá para afirmar o ganho.

## 4 - FEATURE ENGINEERING — (engenharia de atributos)

📖 **O que é:** criar variáveis novas (*features*, atributos ou variáveis independentes) a partir das existentes, para extrair mais do que os dados já sabem. Inclui transformar matematicamente as colunas e **combinar** colunas (razões, diferenças, contagens).

🧠 **Exemplos que funcionaram no portfólio**
- **Razões:** uso do limite = saldo ÷ limite; ticket médio = compras ÷ número de compras (Segmentação).
- **Datas:** dia da semana, semana do ano, dias desde a última promoção, feriado (Rossmann).
- **Agregações:** média da loja nas semanas anteriores (Rossmann), número de leitores de um livro (Livros).

⚠️ **Correção de uma ideia comum:** criar muitas variáveis com **poucos dados aumenta** o risco de sobreajuste (o modelo encontra padrões que são só ruído). Crie poucas e com motivo de negócio.

⚠️ **Cuidado com o vazamento:** atributos que usam o alvo (ex.: média de vendas por loja) devem ser calculados **só com o passado ou só com o treino**, de preferência dentro do Pipeline.

🧠 **Em recomendação**
- **Identificador do item:** junte as edições de um mesmo item (no projeto de Livros, obra = título + autor normalizados, em vez do ISBN de cada edição ou só do título).
- **Uma interação por usuário × item:** quando há duplicatas, fique com a maior nota.
- **Catálogo:** só itens com dados suficientes (ex.: 50 ou mais notas), para a similaridade ser confiável.

## 5 - EXPLORANDO AS CORRELAÇÕES

📖 **O que é:** medir **como as variáveis andam juntas**. Isso serve para duas coisas: encontrar as que se relacionam com o alvo e achar **variáveis redundantes** (duas colunas que carregam a mesma informação, as *colineares*).

🧠 **Correlação**
- Vai de **−1 a +1**. **+1:** quando uma sobe, a outra sobe na mesma proporção; **−1:** quando uma sobe, a outra desce; **0:** nenhuma relação **em linha reta ou monotônica**.
- **Pearson** mede relação em **linha reta** e é sensível a outliers.
- **Spearman** usa a ordem dos valores e mede se uma variável **sempre sobe (ou desce)** com a outra, mesmo em curva. É o padrão do portfólio para dados assimétricos.
- Para categóricas, use o **V de Cramér** (de 0 a 1).

![Correlação](imagens/correlacao.svg)

No quarto gráfico, a relação é perfeitamente crescente, mas curva: o Pearson dá 0,85 e o Spearman, 0,95.

⚠️ **Correlação não é causalidade.** E correlação perto de zero não prova que não há relação: pode haver uma relação em U, que nenhum dos dois coeficientes capta.

🧠 **Por que tirar as variáveis redundantes?**
- **Dimensionalidade** é o número de colunas. Quanto mais colunas, mais complexo o modelo e mais dados ele precisa para aprender bem.
- Pela **navalha de Occam** (princípio de preferência, não garantia), entre dois modelos com desempenho parecido, prefira o mais simples: ele costuma generalizar melhor e é mais fácil de explicar.
- Duas colunas com correlação acima de **0,9** (em módulo) dizem quase a mesma coisa; manter só uma simplifica sem perder informação. Em álgebra linear, são vetores quase **linearmente dependentes**.

✅ **O que fazer:** mapa de calor de Spearman, lista dos pares acima de 0,9 e a decisão de quais manter.

## 3 - Transformação

📖 **O que é:** a maioria dos algoritmos aprende melhor com **números na mesma escala**. Muitos usam otimização (derivadas, distâncias), e uma coluna em milhares de reais "grita" mais alto que uma coluna de 0 a 1. Os dois tipos de transformação são **de grandeza** (escala, log) e **de natureza** (respeitar o que o dado é, como o ciclo dos meses).

🧠 **Escalas (os nomes certos)**

| Técnica | O que faz | Quando usar |
|---|---|---|
| **Padronização** (`StandardScaler`, z-score) | Subtrai a média e divide pelo desvio: média 0, desvio 1 | Padrão para modelos lineares, K-Means e redes neurais |
| **Normalização Min-Max** (`MinMaxScaler`) | Leva tudo para o intervalo de 0 a 1; mantém a forma da distribuição | Quando é preciso um intervalo fixo; sensível a outliers |
| **RobustScaler** | Subtrai a mediana e divide pelo IQR | Colunas com outliers fortes |
| **Log** (`log1p`) | Comprime a cauda longa (1, 10, 100 viram 0,7; 2,4; 4,6) | Valores em dinheiro, contagens, alvos assimétricos |

![Escalas](imagens/escalas.svg)

> Modelos de árvore (Random Forest, XGBoost, LightGBM) **não precisam** de escala: eles só comparam "maior ou menor que".

🧠 **Encoding: transformar categorias em números**

| Encoder | Como funciona | Quando usar |
|---|---|---|
| **One-hot** (ou *dummies*) | Uma coluna 0/1 para cada categoria | Categorias sem ordem e com poucos níveis (regra prática: até 6 ou 7) |
| **Ordinal** | Números na ordem certa (P=0, M=1, G=2) | Categorias **com ordem** (escolaridade, tamanho, frio/morno/quente) |
| **Label encoding** | Números arbitrários (azul=0, verde=1, preto=2) | Só para o **alvo** ou em árvores; em modelos lineares cria uma **ordem falsa** |
| **Frequência** | Troca a categoria pela sua frequência | Muitos níveis (cidades, produtos) |
| **Target encoding** | Troca a categoria pela média do alvo naquela categoria | Muitos níveis; **calcule só no treino ou dentro da validação cruzada**, senão é vazamento |
| **Target encoding ponderado** | Mistura a média da categoria com a média geral, pesando pelo tamanho da categoria | Categorias com poucas linhas (evita confiar numa média de 3 casos) |

![Encoding](imagens/encoding.svg)

🧠 **Transformação de natureza (variáveis cíclicas)**

Mês, dia da semana e hora são **ciclos**: dezembro está ao lado de janeiro. Como número (1 a 12), dezembro parece longe de janeiro. Seno e cosseno colocam os meses num círculo:

🧠 **Pipeline**

Um `Pipeline` encadeia todas as etapas (nulos → atributos → escala/encoding → modelo). No `.fit()` cada etapa aprende **só com o treino**; no teste e no app, as mesmas etapas são apenas **aplicadas**. Isso evita vazamento e garante que o app use exatamente o tratamento do treino.

![Pipeline](imagens/pipeline.svg)

🧠 **Matriz usuário × item:** os modelos recebem uma **matriz esparsa** (a maioria das células é vazia: cada pessoa conhece uma fração mínima do catálogo), com 1 onde o usuário tem o item. Qual interação vira 1 (só "gostou", todas as notas ou também as sem nota) é um hiperparâmetro testado no ajuste fino.

## 5 - Treinamento dos Algoritmos de Machine Learning

📖 Todos os modelos treinam com a mesma matriz e são medidos na **validação**: os itens escondidos de cada usuário precisam aparecer entre os N recomendados. Os itens que o usuário já tem **nunca** são recomendados de novo (no v1 dos Livros, o próprio livro escolhido ocupava 1 das 5 vagas).

![Acerto@K](imagens/acerto_k.svg)

| Métrica | O que mede | No exemplo |
|---|---|---|
| **Acerto@N** (*hit rate*) | % de usuários com pelo menos 1 item escondido na lista | sim (2 acertos) |

| **Recall@N** | Dos itens escondidos, quantos a lista encontrou | 2/4 = 50% |
| **NDCG@N** | Como o recall, mas vale mais acertar no topo da lista (1 = acertos todos no topo) | 0,40 |
| **Cobertura** | % do catálogo que aparece em alguma lista | |
| **Popularidade média** | % de usuários que já consumiram os itens recomendados (alta = só best-sellers) | |

| **Wilcoxon pareado** | Teste para a mesma unidade medida duas vezes (antes × depois), sem supor normalidade. |
| **Épsilon²** | Tamanho do efeito do Kruskal-Wallis (0,01 pequeno, 0,08 médio, 0,26 grande). |
